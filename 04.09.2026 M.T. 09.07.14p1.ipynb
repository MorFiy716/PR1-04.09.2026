### Задание 3. Сортировка данных

* Найдите 10 треков с максимальным количеством прослушиваний (stream_count)

```python
top_10_tracks = df.nlargest(10, 'stream_count')
```

```python
print("10 треков с максимальным количеством прослушиваний:")
display(top_10_tracks[['track_name', 'artist_name', 'genre', 'stream_count', 'release_year']])
```

* Какие исполнители и жанры встречаются среди самых прослушиваемых треков?

```python
print("Уникальные исполнители из топ 10:")
print(top_10_tracks['artist_name'].unique())
```

```python
print("Уникальные жанры из топ 10:")
print(top_10_tracks['genre'].unique())
```

### Задание 5. Основные статистические показатели

### Задание 6. Мода

Найдите наиболее часто встречающееся значение (моду):

* Для столбца genre

```python
mode_genre = df['genre'].mode()
print(f"Мода для столбца genre: {mode_genre.values}")
```

* Для столбца release_year

```python
mode_release_year = df['release_year'].mode()
print(f"Мода для столбца release_year: {mode_release_year.values}")
```

### Задание 7. Гистограмма

Постройте гистограмму распределения значений popularity.

```python
plt.figure(figsize=(10, 5))
plt.hist(df['popularity'], bins=30, edgecolor='black')
plt.title('Распределение популярности треков на Spotify')
plt.xlabel('Популярность (popularity)')
plt.ylabel('Количество треков')
plt.grid(True, alpha=0.3)
plt.xticks(range(0, 101, 10))
plt.show()
```

* В каком диапазоне находится основная часть значений?

В диапазоне от 0 до 60

* Есть ли значения, которые встречаются значительно реже остальных?

Да, от 80 до 100

### Контрольные вопросы

1. Что такое DataFrame?
Двумерная табличная структура данных с подписанными строками и столбцами.

2. Что такое признак (Feature)?
Отдельный столбец в наборе данных, описывающий одну характеристику объектов.

3. Чем среднее арифметическое отличается от медианы?
Среднее — сумма значений, делённая на их количество; медиана — значение в середине упорядоченного ряда.

4. Что показывает гистограмма?
Распределение частот числовых данных по интервалам.

5. Для чего нужна фильтрация данных?
Для выборки строк, удовлетворяющих заданному условию.
